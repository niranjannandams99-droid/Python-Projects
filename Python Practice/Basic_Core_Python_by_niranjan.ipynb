```python
#Swap Two Numbers
a = int(input('enter number'))
b = int(input('enter number'))

print(f'original a = {a} ,b = {b}')

a,b =b,a
print(f'swapped a={a} ,b = {b}')
```

```python
#Area of Circle
# area = pi( R^2) --- r = raidus

Radius = float(input('enter number'))

pi = 3.14159265359

area = pi*(Radius **2)

print(f'area of the circle {round(area,2)}')
```

```python
#Area of Rectangle
# area of the Rectangle = length * width 
length = int(input('enter length: '))
width = int(input ('enter width: '))

AOR = length * width

print(f'Area of the Rectangle{AOR} ')
```

```python
#Celsius to Fahrenheit

#f =( c *(9/5) ) + 32

Celsius = float(input ('enter the temp: '))

fahrenheit = (Celsius * (9/5))+ 32
print(f'Celsius {Celsius} °C to fahrenheit {fahrenheit}°F')
```

```python
#Simple Interest
#Write a Python program that calculates the Simple Interest for a student taking 
# a loan of ₹15,000 for their college tuition, at an annual interest rate
# of 6.5%, for a duration of 3 years. Print the final Simple Interest amount.


loan = int(input('Enter loan amount'))
Intrest_rate = float(input ("enter a intrest rate: %"))

si = (loan*Intrest_rate)*1/100

print(f'loan amount {loan} /- and the intrest rate is {Intrest_rate}% and the final simple intrest per year {si} /- ,final simple intrest per 3 years = {si*3} /- ')
```

```python
#BMI Calculator

#Body mass index = weight in kg/height in meters ^2

weight = float(input("enter your weight in KGS"))
height = float(input("enter your height in meters"))

BMI = weight /(height)**2

print(f'ur weight in KGs {weight} and ur height in meters {height} and ur body mass index is {BMI}')
```

```python
#Percentage Calculator
#A store is offering a 15% discount on a jacket that originally costs ₹2,500. Write a Python program to calculate the 
#discount amount (how much money you save) and the final price you have to pay.

item_cost= float(input("enter a iteam cost"))
discount = float(input("enter a discount : %"))

dis_amount = item_cost*discount/100
print(f'your item amount {item_cost} /- and your discount {discount}% and your discount is :-{dis_amount} total amount is {item_cost - dis_amount}')
```

```python
#Age Calculator

DOB = int(input("Enter your date of birth"))
current_year = int (input("enter a current year"))

age = current_year - DOB
print(f'your age is {age}')
```

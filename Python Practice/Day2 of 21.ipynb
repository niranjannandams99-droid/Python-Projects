# Mini Project :
Student Result System

Ask user:

Name

Marks in

Python

SQL

Excel

Statistics

Calculate:

Total

Average

Grade

Pass/Fail



```python
student_name = input("Enter your Name: ")   # Name

print(f'--------Enter your marks---------')

marks = ["PYTHON", "SQL", "EXCEL", "STATISTICS"]   # iteration for all topics inputs
ALL_MARKS = []
for i in marks:
    S_marks = int(input(i))
    ALL_MARKS.append(S_marks)

print(f'Your Total marks is : {sum(ALL_MARKS)}')          # Total_marks
print(f'Your Average Marks is : {(sum(ALL_MARKS)/4)}')     # average


if sum(ALL_MARKS) > 50:                           #Grade over all marks
    if sum(ALL_MARKS) >=95:
        print(f'Your Grade is "A+" ')
    elif sum(ALL_MARKS)>85:
        print(f'Your Grade is "A" ')
    elif sum(ALL_MARKS)>=75:
        print(f'Your Grade is "B+" ')
    elif sum (ALL_MARKS) >=65:
        print(f'Your Grade is "B" ')
    else:
        print(f'Your grade is "C" ')
else:
    print("--------Your Fail ----------")

is_passed = True                                       
for subject, marks in zip(marks,ALL_MARKS):          #which subj fail
    if marks <= 25:
        print(f'Your fail in this {subject} : {marks}')
        is_passed = False

if is_passed:                                       #overall pass or fail 
    print("Result : passed the test")
else:
    print('Result :Fail')
```
